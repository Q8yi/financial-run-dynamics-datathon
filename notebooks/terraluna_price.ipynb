```python
import pandas as pd
import matplotlib.pyplot as plt
ust_price = pd.read_csv("../datasets/processed/erc_20/processed_ustc_price_data.csv")
ust_price.columns
```

```python
ust_price.head()
```

```python
ust_price.shape
ust_price.dtypes
```

```python
ust_price["date"] = pd.to_datetime(ust_price["date"])
ust_price = ust_price.sort_values("date")
```

```python
zoom = ust_price[(ust_price["date"] >= "2022-04-20") & (ust_price["date"] <= "2022-05-12")]

plt.figure(figsize=(10,5))
plt.plot(zoom["date"], zoom["close"], label="UST close")
plt.axhline(1.0, linestyle="--", color="grey", alpha=0.7, label="Peg ($1)")

plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.title("UST price around depeg (daily close)")
plt.legend()
plt.tight_layout()
plt.show()
```
